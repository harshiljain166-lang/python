numpy indexing and slicing(of a matrix)

```python
arr=np.arange(1,31).reshape(6,5)
#arr[row_start:row_stop,col_start:col_stop]
arr
```

```python
arr[0,1]
```

```python
arr[0:2,0:2]
```

```python
arr[3:5,3]
```

```python
arr[2:3,3:4]
```

```python
arr[2:,1:3]
```
